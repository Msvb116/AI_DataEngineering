```python
df = spark.read.csv("C:\\Users\\DELL\\Desktop\\AI_DATA_ENGINEERING\\Data_Sets\\sales_info.csv" , header = True , inferSchema = True , escape = '"')
```

```python
df.show(5)
```

```python
df.sort(asc("C_Year")).show(10)
```

```python
# YoY % = ((Current Year Sales - Previous Year Sales) / Previous Year Sales) * 100
```

```python
#in python we cannot do row level calculation 
# In sql,Python, spark cannot do row level transformations
# column level calculation is not possible
# pandas --> shift function
# sql  ----> lead() & lag()
# pyspark -->  lead() & lag()
```

```python
windowspec = Window.orderBy("C_Year")
```

```python
df = df.withColumn("PY_Sales",
                   lag("T_Sales").over(windowspec))
```

```python
df.show()
```

```python
df = df.withColumn(
    "YoY",
    round(
        ((col("T_Sales") - col("PY_Sales")) / col("PY_Sales")) * 100,
        2
    )
)
```

```python
df.show()
```
