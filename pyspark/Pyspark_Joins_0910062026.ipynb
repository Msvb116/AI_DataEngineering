```python
emp.createOrReplaceTempView("Emp_Tb")
train.createOrReplaceTempView("Train_Tb")
```

```python

spark.sql("""
select a.Title,
       sum(b.Cost) as total_cost
from Emp_Tb a
inner join Train_Tb b
on a.EmployeeID = b.Employee_ID
group by a.Title
order by sum(b.Cost) desc
limit 5
""").show()
```

```python
select a.title, sum(b.cost) from Emp_Tb a inner join Train_Tb b on a.EmployeeID = b.Employee_ID group by a.Title order by sum(b.Cost) desc limit 5
```

```python
# to rename columns 
cust.join(ords, cust['ID'].cast("string") == ords['C_ID'] , "inner") \
    .select(col("NAME").alias("Cust_Name"),col("Address").alias("Cust_Address"),col("OID").alias("Ord_ID"),col("Date").alias("Ord_Date")).show()
```

```python
# i need emp who has spent more on training cost (sales dept) , but i have to see who have spent more than 25 k
#Where clause filter condition
# groupby 
## having condition filter condition
```

```python

emp.join(train , emp["EmployeeID"] == train["Employee_ID"], "inner") \
    .filter(col("Dept") == "Sales")  \
   .groupby(col("Title").alias("Emp_Name")).agg(sum("Cost").alias("Tot_Cost")) \
   .filter(col("Tot_Cost") > 25000) \
    .show(5)
```
