```python
data1.createOrReplaceTempView("Data")
```

```python
spark.sql("select Sub_Category , sum(Sales) Tot_sales , sum(Profit) Tot_profit  \
            , sum(Quantity) Tot_Qty from Data group by Sub_Category order by sum(Sales) desc").show()
```

```python
# i want for every sub category i need total sales , tot qty , tot_ profit 
# want to see resullts tot_qty > 3000
```

```python
spark.sql("select Sub_Category , sum(Sales) Tot_sales , sum(Profit) Tot_profit  \
            , sum(Quantity) Tot_Qty from Data group by Sub_Category  \
             having sum(Quantity) > 3000 order by sum(Sales) desc").show()
```

```python
data1.groupby("Sub_Category").agg(sum("Sales").alias("Tot_Sales"), \
                                 sum("Profit").alias("Tot_Profit"), \
                                 sum("Quantity").alias("Tot_Qty")) \
                            .filter(col("Tot_Qty") > 3000) \
                            .sort(desc("Tot_Sales")).show()
```

```python
# Need only East region Records for east region you should get me all sub category , tot_sales, avg_sales also sort
# with avg sales highest to lowest
# give me those sub category sales more than 50,000 (filter on aggregation)
```

```python
data1.filter(col("Region") == "East").groupby("Sub_Category") \        # where clause in sql
                                     .agg(sum("Sales").alias("Tot_Sales"), \
                                          avg("Sales").alias("Avg_Sales")) \
                                    .filter(col("Tot_Sales")>50000) \    # having clause in sql
                                    .sort(desc("Avg_Sales")).show()
```

```python
# always remember the order
```

```python
data1.filter(col("Region") == "East") \
    .groupby("Sub_Category") \
    .agg(
        sum("Sales").alias("Tot_Sales"),
        avg("Sales").alias("Avg_Sales")
    ) \
    .filter(col("Tot_Sales") > 50000) \
    .sort(desc("Avg_Sales")) \
    .show()
```

```python
spark.sql("""select sub_category,sum(sales) tot_sales, avg(sales) avg_sales from Data where region = 'East' group by Sub_Category   
             having sum(sales) > 5000 order by avg(sales) desc""").show(20)
```

```python
data1.groupby("Sub_Category", "Region") \
    .agg(
        sum("Sales").alias("Total_Sales"),
        avg("Sales").alias("Average_Sales")
    ) \
    .filter(
        (col("Region") == "East") &
        (col("Total_Sales") > 50000)
    ) \
    .sort(desc("Average_Sales")) \
    .show(20)
```

```python
data1.filter(col("Region") == "East") \
    .groupby("Sub_Category") \
    .agg(
        sum("Sales").alias("Tot_Sales"),
        avg("Sales").alias("Avg_Sales")
    ) \
    .sort(desc("Avg_Sales")) \
    .filter(col("Tot_Sales") > 50000) \
    .show()
```
