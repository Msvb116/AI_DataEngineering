```python
from abc import ABC , abstractmethod
```

```python
class KSRDetails(ABC):
    def __init__(self, data):
        self.data = data

    @abstractmethod
    def trainers_details(self):
        pass

    @abstractmethod
    def trainers_qualification(self):
        pass

    @abstractmethod
    def trainer_salary(self):
        pass
```

```python
class KSRImplementationClass(KSRDetails):
    def trainers_details(self):
        print("All trainers are professional")
        
    def trainers_qualification(self):
        print("All trainers have masters degree")
        
    def trainer_salary(self):
        return 2000*100
```
