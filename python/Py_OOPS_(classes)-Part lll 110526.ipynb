```python
class data_cleaning:
    def __init__(self,data):
        self.data = data
    def volume_chk(self):
        return self.data.shape
    def sample_data(self):
        return self.data.head(5)
    def data_typed(self):
        return self.data.dtypes
    def row_count(self):
        print("The No of rows in given data set:" , self.data.shape[0])
    def column_count(self):
        print("The No of columns in given data set:" , self.data.shape[1])
    def missing_values(self):
        return self.data.isna().sum()
    def missing_value_percentage(self):
        return (self.data.isna().sum() / self.data.shape[0]) * 100
    def column_list(self):
        return list(self.data.columns)

        
```
