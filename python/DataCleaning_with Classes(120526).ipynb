```python
import pandas as pd
import numpy as np
import warnings
warnings.filterwarnings("ignore")
```

```python
cust_data = pd.read_excel("C:\\Users\\DELL\\Desktop\\AI_DATA_ENGINEERING\\Data_Sets\\Cust_Data.xlsx")
```

```python
class cleaning_data:
    def __init__(self,data):
        self.data = data
    def volume_check(self):
        return self.data.shape
    def sample_data(self,N):
        return self.data.head(N)
    def row_count(self):
        print("The No of rows in given data set:" , cust_data.shape[0])
    def column_count(self):
        print("The No of Columns in given data set:" , cust_data.shape[1])
    def column_list(self):
        return list(cust_data.columns)
    def numerical_columns(self):
        return list(cust_data.select_dtypes(include = np.number).columns)
    def cartergial_columns(self):
        return list(cust_data.select_dtypes(exclude = np.number).columns)
    def drop_duplicates(self):
        self.data = self.data.drop_duplicates()
    def drop_nulls(self,column_name):
        self.data = self.data.dropna(subset = [column_name])
    def missing_value(self):
        return self.data.isna().sum()
    def missing_value_percentage(self):
        return (self.data.isna().sum() / self.data.shape[0]) * 100
    def impute_missing_val_with_mean(self,col_name):
        self.data['col_name'] = self.data['col_name'].fillna(self.data['col_name'].mean()).astype("int")
    def impute_missing_val_with_mode(self,col_name):
        self.data['col_name'] = self.data['col_name'].fillna(self.data['col_name'].mode()[0])
    def impute_missing_val_with_others(self,col_name):
        self.data['col_name'] = self.data['col_name'].fillna("Guest/Na")
    def phonenumber_masking(self, column_name):
        def mask_logic(b):
            b = str(b).split('.')[0] # to remove floating / decimal points any
            if b == '0' or b == '0000000000':
                return 'XXXXXXXXXX'
            else:
                returnb[0:3] + 'XXXXX' + b[-2:]
        self.data[column_name] = self.data[column_name].apply(mask_logic)
```
