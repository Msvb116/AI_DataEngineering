```python
cust_data = cust_data.dropna() # which ever row had null value comple row will be dropped 
```

```python
cust_data = cust_data.dropna(subset = 'CustID') # this will drop only if custid had Na value
```

```python
cust_data.head(10)
```

```python
cust_data.shape
```

```python
cust_data = cust_data.drop_duplicates()
```

```python
cust_data.shape
```

```python
cust_data['CustAge'] = cust_data['CustAge'].fillna(cust_data['CustAge'].mean()).astype("int")
```

```python
cust_data.head(10)
```

```python
cust_data['CustLoc'] = cust_data['CustLoc'].fillna(cust_data['CustLoc'].mode()[0]) #[0] is index
```

```python
cust_data.head(10)
```

```python
cust_data['CustName'] = cust_data['CustName'].fillna("Guest/Na")
```

```python
cust_data.head(10)
```

```python
cust_data['CustID'] = cust_data['CustID'].astype("int")
```
