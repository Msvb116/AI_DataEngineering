```python
df = data.groupby('broker_id')[['firm_sales','global_sales']].sum().reset_index() # assign to new df (dataframe )
```

```python
df.head(5)
```

```python
df['loyality_percentage'] = (df['firm_sales']/ df['global_sales']) * 100
```

```python
df.shape
```

```python
df.head(5)
```

```python
df.sort_values('loyality_percentage', ascending = False).head(18)
```

```python
# function to add loyal category as per the percentage
def loyality_check(loyality_percentage):
    if loyality_percentage == 100:
        return "Loyal Brokers"
    elif loyality_percentage >= 85 and loyality_percentage < 100:
        return "SteadyFast Brokers"
    elif loyality_percentage >= 50 and loyality_percentage < 85:
        return "Reasonabily Devoted Brokers (Moderate)"
    elif loyality_percentage < 50 and loyality_percentage > 0:
        return "Less Inclined Brokers"
    elif loyality_percentage == 0:
        return "Non Comitted brokers"
    else:
        return "Stagent Brokers"
```

```python
# here we are adding new column and applying function to it 
df['Loyality_category']= df['loyality_percentage'].apply(loyality_check)
```
