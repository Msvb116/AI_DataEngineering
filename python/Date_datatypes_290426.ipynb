```python
# step 1 - Change the data types to date object to date time
```

```python
data['OrderDate'] = pd.to_datetime(data['OrderDate'])
```

```python
data.dtypes
```

```python
data['Year'] = data['OrderDate'].dt.year
data['Month'] = data['OrderDate'].dt.month
data['day'] = data['OrderDate'].dt.day
```

```python
data.dtypes
```

```python
data.head(3)
```

```python
data['DayName'] = data['OrderDate'].dt.day_name()
```

```python
data.head(3)
```

```python
data.groupby("DayName")[["Sales"]].sum().reset_index().sort_values("Sales" , ascending = False)
```
