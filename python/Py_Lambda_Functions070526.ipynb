```python
# lambda function
```

```python
addition_L = lambda a, b : a + b # one line code
```

```python
addition_L(11,7)
```

```python
txn_grp_L = lambda sales:"Silver" if sales < 5000 else "Gold"
```

```python
txn_grp_L(5600)
```

```python
# Lambda functions 
# it is a small anynomous function 
# It is just Single line 
# Syntax : fun --> lambda arguments: expression
# good for performing short operations / data manipulation

#===============================================
# Cannot be used for complex transformations
# looses Redability
```

```python
def add_L(a,b):   # Normal function
    c = a + b
    return c
```

```python
add_L = lambda a,b :  a + b  # lambda function 
```

```python
def txn_grp(sales):
    if sales < 5000:
        return "Silver"
    else:
        return "Gold" 
```

```python
txn_grpL = lambda sales: "Silver" if sales < 5000 else "Gold"
```

```python
def txn_grp(sales):
    if sales < 5000:
        return "Silver"
    elif sales < 10000:
        return "Gold"
    else:
        return "Platinum" 
```

```python
txn_grpL = lambda sales : "Silver" if sales < 5000 else ("Gold" if sales < 10000 else "Platinum")
```
