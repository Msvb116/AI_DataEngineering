```python
data1 = pd.read_csv("C:\\Users\\DELL\\Desktop\\AI_DATA_ENGINEERING\\Data_Sets\\Sample_Store_1.csv", encoding = "ISO-8859-1")
```

```python
data1.head(5)
```

```python
df  =data1[["Segment", "Region" , "Category" , "Sales" , "Profit"]]
```

```python
df
```

```python
df['Sales_Grp'] = np.where(df['Sales'] < 5000 , 'Bronze', 'Silver') # this condition only have true or false 
```

```python
df['Sales_Grp'] = np.where(df['Sales'] < 5000 , 'Bronze', np.where(df['Sales'] < 10000 , "Silver" ,"Gold"))
```

```python
df .head(5)
```

```python
df.groupby("Sales_Grp")[["Segment"]].count()
```

```python
df['Sales_Grp'] = np.select([df['Sales'] < 5000 , df['Sales'] < 10000 , df['Sales'] > 10000 ] ,  ["Bronze","Silver","Gold"])
```

```python
df.head(10)
```

```python
df['Sales_Grp'] = np.select(
    [
        df['Sales'] < 5000,
        (df['Sales'] >= 5000) & (df['Sales'] < 10000),
        df['Sales'] >= 10000
    ],
    [
        "Bronze",
        "Silver",
        "Gold"
    ],
    default="Unknown"
)
```
