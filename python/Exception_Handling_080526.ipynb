```python
try:
    import panda as pd
except:
    print("There is typo Error")  
```

```python
try:
    a = [1,2,5,6,7,9,10]
    print(a[7])
except:
    print("You are trying to access out side list")  
```

```python
try:
    num = 10
    deno = 0
    res = num/deno
    print(res)
except:
    print("You Cann't Divide any No by Zero")  
```

```python
try:
    a = 10
    b = 6
    print(a)
    print(b)
except:
    print("Check The Alignment")  
```

```python
try:
    a = 40
    if a < 35:
        Print("Fail")
    else:
        Print("Pass")
except:
    print("Pls Check the indentation Error")
```

```python
try:
    num = [1,2,3,4]
    a = 10
    b = 4
    c = a/b
    d = num[5]
    print("Acces Element: ",d)
    print("Divide by 2 No's ", c)
except:
    print("There is code error, While accessing Index")
```

```python
try:
    import pandas as pd
    num = [1,2,3,4]
    a = 10
    b = 0
    c = a/b
    d = num[0]
    print("library imported")
    print("Acces Element: ",d)
    print("Divide by 2 No's ", c)
except IndexError:
    print("There is code error, While accessing Index")
except ZeroDivisionError:
    print("There is code error, While Division")
except ModuleNotFoundError:
    print("There is code error, Module not found Error")    
```

```python
try:
    import pandass as pd
    data = pd.read_excel("C:\\Users\\DELL\\Desktop\\AI_DATA_ENGINEERING\\Data_Sets\\student_infos.xlsx")
    num = [1,2,3,4]
    a = 10
    b = 0
    c = a/b
    d = num[0]
    print("library imported")
    print("Acces Element: ",d)
    print("Divide by 2 No's ", c)
except Exception as e:
    _, _, tb = sys.exc_info()
    summary = traceback.extract_tb(tb)
    line_no = summary[-1].lineno
    print(e)
    print(f"Error occur at {line_no}")
```
