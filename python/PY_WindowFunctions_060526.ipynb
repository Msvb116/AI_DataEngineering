### window Functions
### Functions where logic is applied for certain columns to derive new column

```python
data.head(10)
```

```python
data['Rank'] = data['MARKS'].rank(method = 'min', ascending = True)
```

```python
data.head(10)
```

```python
# here we are getting decimal ranking do casting 
# min --> rank function
# Dense rank - rank will not be skipped
# problemm with sense rank --> not applicable for ranking systems ex: olyampics
# Row number we didnt see duplicate ranks we will get consecutive ranks 
```

```python
data['Rank'] = data['MARKS'].rank(method = 'min', ascending = False).astype('int')
data['DENSE'] = data['MARKS'].rank(method = 'dense', ascending = False).astype('int')
data['ROW_NUMBER'] = data['MARKS'].rank(method = 'first', ascending = False).astype('int')
```

```python
data.head(10)
```

```python
data.sort_values("Rank", ascending = True) # rank function will skip rank incase of same rank
```

### YoY % = ((Current Year Sales - previous Year sales)/previous Year sales ) * 100

```python
sales = sales.sort_values("C_Year" , ascending = True)
```

```python
sales['PY_Sales'] = sales['T_Sales'].shift(1)  # shift will create 1 step lag as follows
```

```python
sales.head(10)
```

```python
sales['YoY'] = ((sales['T_Sales'] - sales['PY_Sales'])/ sales['PY_Sales']) * 100
```

```python
sales.head(10)
```
