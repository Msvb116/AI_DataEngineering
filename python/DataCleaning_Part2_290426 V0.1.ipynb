```python
import pandas as pd
import datetime
import warnings
warnings.filterwarnings('ignore')
from datetime import date
```

```python
data = pd.read_csv("C:\\Users\\DELL\\Desktop\\AI_DATA_ENGINEERING\\Data_Sets\\EmployeeDataToClean.csv" , encoding = "ISO-8859-1")
```

```python
data.head(5)
```

```python
def phn_masking(a):
    a = str(a)
    if a == '0':
        return "XXXXXXXXXX"
    else:
        result = a[0:3] + "XXXX" + a[-2:]
    return result    
```

```python
def f_materialstatus(a):
    if a == 'M':
        return 1
    else:
        return 0
```

```python
def f_genderstatus(a):
    if a == 'M':
        return 1
    else:
        return 0
```

```python
# data Cleaning Steps
data['PhoneNumber'] = data['PhoneNumber'].fillna(0000000000)
data['PhoneNumber'] = data['PhoneNumber'].astype("object")  # here decimal values will come to overcome this we use int64 thpe
data['PhoneNumber'] = data['PhoneNumber'].astype("int64")   # here warnings are coming to overcome this keep the import warnings syntax
data['PhoneNumber'] = data['PhoneNumber'].apply(phn_masking)
data["Is_Married"] = data['MaritalStatus'].apply(f_materialstatus)
data["Is_Male"] = data['Gender'].apply(f_genderstatus)
data = data.drop(["NationalIDNumber","LoginID","MaritalStatus", "Gender", "CurrentFlag" , "rowguid"] , axis = 1)
data["Salary"] = data["Salary"].fillna(data["Salary"].mean())
data['Dept']  = data['Dept'].str.capitalize()
data['BirthDate'] = pd.to_datetime(data['BirthDate'])
#data['HireDate'] = pd.to_datetime(data['HireDate'])
data['HireDate'] = pd.to_datetime(data['HireDate'], errors='coerce')
data['Age'] = date.today().year - data['BirthDate'].dt.year
```
